```python
# ============================================
# PROJECT: TASK TRACKER
# PURPOSE: Add, view, complete, and delete tasks
# ============================================


# STEP 1: Import the JSON module
# JSON helps us save Python data into a file
# and read that data again later.

import json


# STEP 2: Define the name of our JSON file
# All tasks will be saved inside this file.

FILE_NAME = "tasks.json"


# STEP 3: Create a function to load saved tasks
# This function reads tasks from tasks.json.

def load_tasks():

    # Try to open the JSON file.
    try:

        # Open the file in read mode ("r").
        with open(FILE_NAME, "r") as file:

            # Convert JSON data into a Python list.
            tasks = json.load(file)

            # Return the loaded list of tasks.
            return tasks

    # If the file does not exist, handle the error.
    except FileNotFoundError:

        # Start with an empty list if no file exists.
        return []

    # Handle an empty or invalid JSON file.
    except json.JSONDecodeError:

        # Print a warning message.
        print("Warning: The task file is empty or invalid.")

        # Return an empty list.
        return []


# STEP 4: Create a function to save tasks
# This function writes tasks into tasks.json.

def save_tasks(tasks):

    # Open the file in write mode ("w").
    # If the file does not exist, Python creates it.
    # If it exists, its old contents are replaced.

    with open(FILE_NAME, "w") as file:

        # Convert the Python list into JSON format.
        # indent=4 makes the file easier to read.
        json.dump(tasks, file, indent=4)


# STEP 5: Create a function to add a new task

def add_task(tasks):

    # Ask the user to enter a task title.
    title = input("Enter task title: ").strip()

    # Check whether the title is empty.
    if title == "":

        # Display an error message.
        print("Error: Task title cannot be empty.")

        # Stop this function without adding a task.
        return

    # Create a dictionary to store the new task.
    # Initially, the task is not completed.
    task = {
        "title": title,
        "done": False
    }

    # Add the new task dictionary to the list.
    tasks.append(task)

    # Save the updated list into the JSON file.
    save_tasks(tasks)

    # Inform the user that the task was added.
    print("Task added successfully!")


# STEP 6: Create a function to view all tasks

def view_tasks(tasks):

    # Check whether the task list is empty.
    if len(tasks) == 0:

        # Tell the user that there are no tasks.
        print("No tasks found.")

        # Stop this function.
        return

    # Print a heading.
    print("\n========== YOUR TASKS ==========")

    # Loop through every task in the list.
    # enumerate() gives each task a number.
    # start=1 makes numbering begin at 1.
    for number, task in enumerate(tasks, start=1):

        # Check whether the task is completed.
        if task["done"] == True:

            # Set its status to Done.
            status = "Done"

        else:

            # Otherwise, its status is Pending.
            status = "Pending"

        # Print the number, title, and status.
        print(f"{number}. {task['title']} [{status}]")


# STEP 7: Create a function to complete a task

def complete_task(tasks):

    # If there are no tasks, nothing can be completed.
    if len(tasks) == 0:

        # Display a message.
        print("No tasks available to complete.")

        # Stop this function.
        return

    # First, display all available tasks.
    view_tasks(tasks)

    # Ask the user for the task number.
    choice = input("Enter task number to complete: ").strip()

    # Check whether the input contains only digits.
    if not choice.isdigit():

        # Display an error for invalid input.
        print("Error: Please enter a valid task number.")

        # Stop this function.
        return

    # Convert the input string into an integer.
    number = int(choice)

    # Check whether the number is within the valid range.
    if number < 1 or number > len(tasks):

        # Display an error for an invalid task number.
        print("Error: Task number does not exist.")

        # Stop this function.
        return

    # Convert the displayed number to a list index.
    # Users count from 1, but Python indexes start at 0.
    task = tasks[number - 1]

    # Check whether the task is already completed.
    if task["done"] == True:

        # Inform the user that it is already completed.
        print("This task is already completed.")

        # Stop this function.
        return

    # Mark the selected task as completed.
    task["done"] = True

    # Save the updated task list.
    save_tasks(tasks)

    # Display a success message.
    print("Task completed successfully!")


# STEP 8: Create a function to delete a task

def delete_task(tasks):

    # Check whether the task list is empty.
    if len(tasks) == 0:

        # Display a message.
        print("No tasks available to delete.")

        # Stop this function.
        return

    # Display the current tasks.
    view_tasks(tasks)

    # Ask the user which task to delete.
    choice = input("Enter task number to delete: ").strip()

    # Check whether the input is a number.
    if not choice.isdigit():

        # Display an error message.
        print("Error: Please enter a valid task number.")

        # Stop this function.
        return

    # Convert the input into an integer.
    number = int(choice)

    # Make sure the selected task exists.
    if number < 1 or number > len(tasks):

        # Display an error message.
        print("Error: Task number does not exist.")

        # Stop this function.
        return

    # Remove the selected task from the list.
    # number - 1 converts the task number into an index.
    removed_task = tasks.pop(number - 1)

    # Save the list after deleting the task.
    save_tasks(tasks)

    # Show the name of the deleted task.
    print(f"Deleted task: {removed_task['title']}")


# STEP 9: Create the main function
# This function controls the whole program.

def main():

    # Load previously saved tasks when the program starts.
    tasks = load_tasks()

    # Keep displaying the menu until the user exits.
    while True:

        # Display the main menu.
        print("\n========== TASK TRACKER ==========")
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Complete Task")
        print("4. Delete Task")
        print("5. Exit")

        # Ask the user to select an option.
        choice = input("Enter your choice: ").strip()

        # Option 1: Add a task.
        if choice == "1":

            # Call the add_task function.
            add_task(tasks)

        # Option 2: View all tasks.
        elif choice == "2":

            # Call the view_tasks function.
            view_tasks(tasks)

        # Option 3: Complete a task.
        elif choice == "3":

            # Call the complete_task function.
            complete_task(tasks)

        # Option 4: Delete a task.
        elif choice == "4":

            # Call the delete_task function.
            delete_task(tasks)

        # Option 5: Exit the program.
        elif choice == "5":

            # Display a goodbye message.
            print("Thank you for using Task Tracker!")

            # Stop the loop.
            break

        # Handle any other menu choice.
        else:

            # Tell the user to select a valid option.
            print("Invalid choice! Please enter 1, 2, 3, 4, or 5.")


# STEP 10: Start the program
# This condition checks whether this file is
# being run directly by Python.

if __name__ == "__main__":

    # Call the main function to start the program.
    main()
```
